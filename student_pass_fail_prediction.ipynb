**Classification Problem :**

Problem statement : Predict if a student is Pass(1) or Fail(0) based on
- Hours studied
- Attendance %

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
x = torch.tensor([
    [1, 40],
    [2, 50],
    [3, 60],
    [4, 65],
    [5, 75],
    [6, 85]
], dtype=torch.float32
)

y = torch.tensor([[0], [0], [0], [1], [1], [1]], dtype=torch.float32)

# Apply normalization only for x values and values of y are already normalized(in b/w 0, 1)
x = x /x.max(dim=0).values
```

```python
x
```

```python
y
```

```python
# Defining model
model = nn.Sequential(
  nn.Linear(2, 8), # 2 i/p neurons & 8 hidden neurons
  nn.ReLU(), # Hidden layer, hence using ReLU activation function
  nn.Linear(8, 1), # 8 hidden neurons, 1 output neuron (fail/pass ; hence 1 neuron)
  nn.Sigmoid() # It is classification, hence using Sigmoid in o/p layer (Sigmoid range [0, 1] ; either fail or pass)
)
```

```python
# Loss & optimizer

criterion = nn.BCELoss() # Classification problem, Hence Binary Cross Entropy(Possibility of outcome is either 0 or 1)
optimizer = optim.Adam(model.parameters(), lr=0.01)
```

```python
# Training

for epoch in range(500):
  pred = model(x)
  loss = criterion(pred, y)

  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

  if epoch % 50 == 0:
    print(f"Epoch {epoch}, Loss {loss.item():.4f} ")
```

```python
# Test Prediction

# Identifying prob of a student who studied 4 hours with 70% attendance
test = torch.tensor([4, 70], dtype=torch.float32)
test = test / x.max(dim=0).values

prob = model(test)

print("Pass probability is : ", prob.item())
print("Prediction :", True if prob > 0.5 else False)
```
